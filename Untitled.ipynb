```python
total_transactions = len(df)
fraud_count = (df['Class'] == 1).sum()
noraml_count = (df['Class'] == 0).sum()
```

```python
fraud_rate = fraud_count / total_transactions * 100
total_amount = df['Amount'].sum()
fraud_amount = df.loc[df['Class'] == 1,'Amount'].sum()
fraud_amount_share = fraud_amount / total_amount * 100 
```

```python
print("Total Transaction : ",total_transactions) 
print("Fraud Transactions : ", fraud_count)
print("Normal Transaction : ", noraml_count)
print(f"Total transaction amount :{total_amount:.2f}")
print(f"Fraud transaction amount : {fraud_amount:.2f}")
print(f"Fraud amount share :{fraud_amount_share:.4f}%")
```

```python
duplicate_rows = df.duplicated().sum()
print("Total duplicate rows : ", duplicate_rows)
duplicate_by_class = (
    df[df.duplicated(keep=False)].groupby('Class').size()
)
print("\n Duplicate-involved rows by class :")
print(duplicate_by_class)
```

```python
duplicate_count = df.duplicated().sum()
total_rows = len(df) 
duplicate_percentage = duplicate_count / total_rows *100

print("Total rows : ", total_rows)
print("Duplicate rows :", duplicate_count)
print(f"Duplicate percentage :{duplicate_percentage:.4f}%")
```

```python
duplicate_rows = df[df.duplicated(keep=False)]
print("Duplicate - involved rows by class : ")
print(duplicate_rows['Class'].value_counts().sort_index())
print("Exact duplicate rows excluding first occurrence : ")
print(df.loc[df.duplicated(),"Class"].value_counts().sort_index())
```

```python
class_summary = df["Class"].value_counts().sort_index()
class_percentage = (
    df["Class"].value_counts(normalize=True).sort_index()*100
)
summary = pd.DataFrame({
    "transaction_Count": class_summary, 
    "Percentage" : class_percentage
})
summary.index = ["Normal (0)", "Fraud(1)"]
display(summary.round(4))
```

```python
plt.figure(figsize=(9,5))
plt.boxplot([
    df.loc[df["Class"] == 0, "Amount"], 
    df.loc[df["Class"] == 1, "Amount"]], 
tick_labels=["Normal", "Fraud"],showfliers=False
)

plt.title("Transaction Amount Distribution : Normal vs Fraud")
plt.ylabel("Transaction Amount ")
plt.xlabel("Transaction Class")
plt.grid(axis="y", alpha=0.3)
plt.show()
```

```python
df["Time_Interval"] = ((df["Time"] // 21600).astype(int))
time_analysis = (
    df.groupby("Time_Interval").agg(Total_Transactions=("Class","size"), 
     Fraud_Transactions = ("Class","sum")))
time_analysis["Fraud_Rate_Percent"] = (time_analysis["Fraud_Transactions"] / time_analysis["Total_Transactions"] * 100)
display(time_analysis.round(4))
```

```python
plt.figure(figsize=(10,5))
plt.bar(
    time_analysis.index.astype(str),time_analysis['Fraud_Rate_Percent'])
plt.title("Fraud Rate by 6 - Hour Dataset Interval")
plt.xlabel("6- Hour Interval Index")
plt.ylabel("Fraud Rate (%)")
plt.grid(axis='y',alpha=0.3)
plt.show()
```
